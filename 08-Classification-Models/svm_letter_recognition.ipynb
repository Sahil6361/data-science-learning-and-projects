# Support Vector Machine: Letter Recognition

## Problem Statement
Classify handwritten/printed letter images into their correct alphabet class from extracted pixel-based features.

## Dataset
**Letter Recognition** dataset (`letterdata.csv`) -- numeric features derived from letter images, with the target letter class.

## Approach
An SVM classifier trained on the feature set and evaluated for classification accuracy.

## Conclusion
Demonstrates SVM's effectiveness on a high-dimensional, multi-class image-feature classification problem.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import svm
```

```python
data=pd.read_csv('letterdata.csv')
```

```python
#x=np.array(data)[:,1:16]
#y=np.array(data)[:]
x,y=np.array(data)[:,1:16],np.array(data.letter)[:] # instaed of writing in 2 line we wrote in single line
```

```python
#split data
x_train=x[:16000,:]
x_test=x[16001:,:]
y_train=y[:16000]
y_test=y[16001:,]
```

```python
model=svm.SVC(C=3)
```

```python
#C is penalty for misclassification
model.fit(x_train,y_train)
```

```python
predict=model.predict(x_test)
```

```python
from sklearn import metrics
print("Confusion Matrix")
c=metrics.confusion_matrix(y_test,predict)
```

```python
print(c)
```
