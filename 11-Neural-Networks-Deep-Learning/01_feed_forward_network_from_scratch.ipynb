# Feed-Forward Neural Network From Scratch (XOR)

## Problem Statement
Build a minimal 2-2-1 feed-forward neural network using only NumPy (no ML framework) and train it to learn the XOR function -- a classic non-linearly-separable problem that a single-layer perceptron cannot solve.

## Dataset
The 4-row XOR truth table (synthetic, not an external file).

## Approach
Implemented sigmoid activation and its derivative by hand, random weight/bias initialization, a manual feed-forward pass, and a full backpropagation + gradient descent training loop (10,000 epochs).

## Conclusion
After training, the network's predictions closely approximate the XOR truth table (0, 1, 1, 0), demonstrating that a hidden layer is what allows the network to learn a non-linear decision boundary.


```python
#creates a network
#2 input
#one output
```

```python
import numpy as np
```

```python
#define activation function
#sigmoid function
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
def sigmoid_derivation(x):
    return x*(1-x)
```

```python
#input data
x=np.array([[0,0],
          [0,1],
          [1,0],
          [1,1]])
```

```python
y=np.array([[0],[1],[1],[0]])  # fixed: was mixing a scalar 0 with lists, now a proper (4,1) column
```

```python
weights_input_hidden=np.random.rand(2,2)
weights_hidden_output=np.random.rand(2,1)
```

```python
#bias
bias_hidden=np.random.rand(1,2)
bias_output=np.random.rand(1,1)
```

```python
def feedforward(x):
    hidden_layer_input = np.dot(x,weights_input_hidden)+bias_hidden
    hidden_layer_output = sigmoid(hidden_layer_input)
    output_layer_input = np.dot(hidden_layer_output,weights_hidden_output)+bias_output  # fixed: was 'bias'
    output = sigmoid(output_layer_input)  # fixed: was 'output_layer'
    return hidden_layer_output,output
```

```python
# Training loop: backpropagation + gradient descent on the XOR problem
learning_rate = 0.5
epochs = 10000

for epoch in range(epochs):
    hidden_layer_output, output = feedforward(x)

    # Backpropagation
    error = y - output
    d_output = error * sigmoid_derivation(output)

    error_hidden = d_output.dot(weights_hidden_output.T)
    d_hidden = error_hidden * sigmoid_derivation(hidden_layer_output)

    # Update weights and biases
    weights_hidden_output += hidden_layer_output.T.dot(d_output) * learning_rate
    weights_input_hidden += x.T.dot(d_hidden) * learning_rate
    bias_output += np.sum(d_output, axis=0, keepdims=True) * learning_rate
    bias_hidden += np.sum(d_hidden, axis=0, keepdims=True) * learning_rate

    if epoch % 2000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss:.4f}")
```

```python
# Final predictions after training
_, final_output = feedforward(x)
print("\nInput:\n", x)
print("\nPredicted output (should approximate XOR: 0,1,1,0):\n", final_output.round(3))
```
