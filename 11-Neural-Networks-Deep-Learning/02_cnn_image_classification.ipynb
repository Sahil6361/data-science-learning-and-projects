```python
# Download the 'punkt' tokenizer if not already downloaded.
# Wrapped in try/except: if Kaggle's internet setting is off, this notebook
# still runs -- word_tokenize falls back to a simple .split() below instead of crashing.
try:
    nltk.download('punkt', quiet=True)
    nltk.download('punkt_tab', quiet=True)
    NLTK_READY = True
except Exception as e:
    print("Could not download NLTK data (no internet?) -- will fall back to simple tokenization.", e)
    NLTK_READY = False
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
```

```python
documents=['I love dogs','I love cats and dogs']
```

```python
#tokenization
if NLTK_READY:
    tokens=[nltk.word_tokenize(doc.lower()) for doc in documents]
else:
    tokens=[doc.lower().split() for doc in documents]  # fallback: simple whitespace tokenization
```

```python
print(tokens)
```

```python
#bags of words
vectorizer=CountVectorizer()
x=vectorizer.fit_transform(documents)
print("Bag of words:",x.toarray())
print("Feature Names:",vectorizer.get_feature_names_out())
```

```python
#TF-IDF
tfidf=TfidfVectorizer()
x_tfidf=tfidf.fit_transform(documents)
print("TF-IDF",x_tfidf.toarray())
print("Feature names:",tfidf.get_feature_names_out())
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM,Dense
```

```python
#sample data
data=np.array([i for i in range(100)])
```

```python
sequence_lenght=10
```

```python
def create_sequence(data,sequence_lenght):
    sequences=[]
    labels=[]
    for i in range(len(data)-sequence_lenght):
        sequences.append(data[i:i+sequence_lenght])
        labels.append(data[i+sequence_lenght])
    return np.array(sequences),np.array(labels)
```

```python
x,y=create_sequence(data,sequence_lenght)
```

```python
x=x.reshape((x.shape[0],x.shape[1],1))
```

```python
train_size=int(len(x)*0.8)
x_train,x_test=x[:train_size],x[train_size:]
y_train,y_test=y[:train_size],y[train_size:]
```

```python
model=Sequential()
model.add(LSTM(50,activation='relu',input_shape=(sequence_lenght,1)))
model.add(Dense(1))
```

```python
model.compile(optimizer='adam',loss='mse')
```

```python
model.fit(x_train,y_train,epochs=200,validation_data=(x_test,y_test))
```

```python
predictions=model.predict(x_test)
```

```python
print("Prediction:",predictions[:5].flatten())
print("Actual:",y_test[:5])
```
