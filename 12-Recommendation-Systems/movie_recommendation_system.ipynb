# Movie Ratings Analysis + Market Basket Association Rules

## Problem Statement
Two related recommendation-system techniques in one notebook: (1) analyze movie ratings to find highly-rated and frequently-rated movies, and (2) mine market-basket transaction data for association rules ("customers who buy X also buy Y").

## Dataset
Note: the original ratings.csv/movies.csv (MovieLens-format) and Market_Basket_Optimisation.csv were **not included** in the uploaded project folder. Synthetic, same-shaped replacements were generated so this notebook runs end-to-end -- swap in the real MovieLens/market-basket datasets for genuine results.

## Approach
Part 1: merge ratings with movie titles, compute per-movie average rating and rating count. Part 2: build a transaction list from basket data and run the Apriori algorithm (`apyori`) to mine association rules by support/confidence/lift.

## Conclusion
Produces a ranked list of top-rated/most-rated movies, plus a set of association rules showing which items tend to be purchased together.

```python
data=pd.read_csv("Market_Basket_Optimisation.csv")
```

```python
data.shape
```

```python
transactions=[]
data_values = data.values  # fixed: was calling .values inside the loop, recomputing it ~150,000 times and making this cell extremely slow
for i in range(0,7501):
    transactions.append([str(data_values[i,j]) for j in range(0,20)])
```

```python
#apriori algorithm
```

```python
from apyori import apriori
```

```python
rules = apriori(transactions, min_support=0.003, min_confidence=0.2, min_lift=3, min_length=2)
results = list(rules)
print(f"Found {len(results)} association rules\n")
for r in results[:10]:
    pair = list(r.items)
    print(f"Rule: {pair[0]} -> {pair[1]}  (support={r.support:.4f})")
```
