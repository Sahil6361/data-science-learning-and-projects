```python
l = [1,2,3,4,5]
#add a new data 200 to list l
#append--- adds a new element to the end of the list it allows single element at a time
l.append(200)
print(l)
```

```python
#add elements 56,34,80 to list l
#extend-->it adds multiple elements to list at its end
l.extend([56,34,80])
print(l)
```

```python
#Add an element 500 to position 4
#insert ---> adds an element to desired index (wherever you want to insert there we can insert)
l.insert(4,500)
print(l)
```

```python
l[4] = 500
print(l)
```

```python
#remove data 500 from list l
#remove---> deletes a specific  data from list
l.remove(500)
print(l)
```

```python
a = 10
if(a>0):
    print("inside if")
    print("a is greater than 0")
    print("a=",a)
print("outside if statement")
```

```python
a = -1
if(a>0):
    print("inside if")
    print("a is greater than 0")
    print("a=",a)
print("outside if statement")
```

```python
a = -1
if(a>0):
    print("inside if")
    print("a is greater than 0")
    print("a=",a)
else:
    print("a is less than 0")
print("outside if statement")
```
