 **membership operator [in, not in ]**


```python
s= 'hello'
#membership operator
#in -- return true if item found in sequence data, else return False
#not in -- return True if item not found in sequence data, else returns False
print('h' in s)
print('h' not in s)

#another method
if 'h' in s:
  print("Present")
else:
  print("Not present")

#another method
for i in range(0,len(s)-1):
  if(s[i] == 'h'):
    print("Present")
    break
else:
  print("Not present")
```

```python
#control flow statement/Decision making statements
#if, elif, else
#Read the number from the user and check if it's equal to zero, greater than zero and less than zero

num = 10  # demo value (Kaggle-safe: was interactive int(input(...)))

if(num==0):
  print(f"The number {num} is equal to zero")
elif num>0:
  print(f"The number {num} is greater than zero")
else:
  print(f"The number {num} is less than zero")
```

```python
#Implement a python code to read age of three people. Determine the oldest and youngest age

a,b,c = 45, 30, 52  # demo values (Kaggle-safe: was interactive map(int, input().split()))
#using min, max function
print("\nThe oldest person (using max) with the age of : ",max(a,b,c))
print("The youngest person (using min) with the age of : ",min(a,b,c))

#To find the oldest
if a>b and a>c:
  grestest = a
elif b>a and b>c:
  grestest = b
else:
  grestest = c

#To find the youngest
if a<b and a<c:
  youngest = a
elif b<a and b<c:
  youngest = b
else:
  youngest = c

print("\nThe oldest person with the age of : ",grestest)
print("The youngest person with the age of : ",youngest)
```
