```python
#creating arrays
#from list
#create a vector
vec = np.array([1,2,3,4,5])
print(type(vec))
```

```python
#create a matrix
#multiple list --> syntax [[],[]] 
mat = np.array([[1,2,3,4],[5,6,7,8]])
print(type(mat))
```

```python
print(mat)
```

```python
#Transpose --> interchanging the rows with colum or vica vesa
# variable_name.T
'''
before
[[1 2 3 4]
 [5 6 7 8]]
after
[[1 5]
 [2 6]
 [3 7]
 [4 8]]
'''
print(mat.T)
```

```python
#Other method to create array
#arange ---- create a vector within specified values
#arrange ----.arange(start,end,step)
v1 = np.arange(0,11,1)
```

```python
print(v1)
```

```python
#converting vector of arrange to matrix
#using reshape(have to check the size is prime or not if yes then we can't do ) -- > 

print(v1.reshape(11,1))
```

```python
v2 = np.arange(0,12,1)
print(v2)
```

```python
#.reshape(m,n)
print(v2.reshape(4,3))
```

```python
#all the metrix element is zero or one -- .ones(5,5) or .ones[5,5]
m1 = np.zeros([5,3])
print(m1)
m2 = np.ones((5,5))
print(m2)
```

```python
#eye metrix --- all the daigonal is one and it's square matrix
m3 = np.eye(4)
print(m3)
```
