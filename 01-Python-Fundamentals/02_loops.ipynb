```python
#Loop  control statements
#break
#continue
#Implement a python code to read numbers from user until he/she presses q. Ask to user to press q every time an input is fed to exit from the application . Compute the average of all number read from user

# Kaggle-safe demo: simulate a user typing 4, 8, 15, 16, then 'q' to stop,
# instead of blocking on real interactive input().
simulated_inputs = ["4", "8", "15", "16", "q"]

l = []
for a in simulated_inputs:
  if a == "q":
    break
  else:
    l.append(int(a))
print("The number are \n",l)

#To find the sum elements of list
total = 0  # renamed from 'sum' -- 'sum' is a Python builtin and shouldn't be overwritten
for i in range(0,len(l)):
  total = total + int(l[i])

p = 1
for i in l:
    p *= i

if len(l) == 0:
  print("\nNo number entered")
else:
  print("the average of sum the number is  : " ,total/len(l))
  print("the average of sum the number is  : " ,total//len(l))
  print("the average of product the number is  : " ,p/len(l))
  print("the average of product the number is  : " ,p//len(l))
```

```python
'''
To a given a list of dictionary where each dictionary contain sales data for a store.
Each dictionary includes the stores name, the month and the sales figure for that month.
Your task is to analyse this data to answer the following question
1. Calulate the total sales for a each store accross all months.
2. Determine the month in which each store had the highest sales.
3. Find the store with the highest total sales and amount of those sales.
'''

sales_data = [
    {"store" : "Store A","month" : "January", "sales" : 15000},
    {"store" : "Store A","month" : "February", "sales" : 17000},
    {"store" : "Store B","month" : "January", "sales" : 12000},
    {"store" : "Store B","month" : "February", "sales" : 13000},
    {"store" : "Store C","month" : "January", "sales" : 20000},
    {"store" : "Store C","month" : "February", "sales" : 19000},
]

def total_sales(sales_data):
    """Q1: total sales per store, across all months."""
    totals = {}
    for row in sales_data:
        store = row['store']
        totals[store] = totals.get(store, 0) + row['sales']
    return totals

def best_month_per_store(sales_data):
    """Q2: the month with the highest sales, for each store."""
    best = {}
    for row in sales_data:
        store, month, sales = row['store'], row['month'], row['sales']
        if store not in best or sales > best[store][1]:
            best[store] = (month, sales)
    return best

def top_store(totals):
    """Q3: the store with the highest total sales, and the amount."""
    top = max(totals.items(), key=lambda kv: kv[1])
    return top

totals = total_sales(sales_data)
print("Total sales per store:", totals)

best = best_month_per_store(sales_data)
print("Best month per store:", best)

top_store_name, top_store_sales = top_store(totals)
print(f"Top store: {top_store_name} with total sales of {top_store_sales}")
```
