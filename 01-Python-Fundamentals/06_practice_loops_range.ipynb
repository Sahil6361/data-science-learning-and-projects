```python
# Kaggle-safe demo: simulate a sequence of typed numbers ending in 'q',
# instead of blocking on real interactive input() forever.
simulated_inputs = [12, 45, 7, 99, 'q']
for num in simulated_inputs:
    print(num)
    print("Press q to exit the application")
    if num == 'q':
        print("Thank you")
        break
```

```python
numbers = []
simulated_inputs = ["14", "22", "8", "q"]  # Kaggle-safe demo: was interactive input() loop
for num in simulated_inputs:
    if (num=='q'):
        break
    else:
        num = int(num)
        numbers.append(num)
if len(numbers)==0:
    print("List is empty")
else:
    print("Numbers are:",numbers)
    print("Average of all numbers: ",sum(numbers)/len(numbers))
```

```python
p=1
for i in numbers:
    if len(numbers)==0:
        print("No numbers in list to compute the product")
        break
    else:
        p=p*i
print("Product of all numbers: ",p)
```

```python
''' Implement a python code to read a n number from user.
Find the odd numbers and even numbers among them.'''
l=[]
odd=[]
even=[]
demo_numbers = [4, 7, 10, 15, 22]  # Kaggle-safe demo values (was interactive input() per iteration)
n = len(demo_numbers)
for i in range(n):
    num = demo_numbers[i]
    l.append(num)
    if (num%2==0):
        even.append(num)
    else:
        odd.append(num)
print("Numbers: ",l)
print("Even Numbers: ",even)
print("Odd Numbers: ",odd)
```

```python
#Decorators
#To add extra functionalities to an existing function
#Rules
#1. Decorator functions need to be nested function
#2. The outer functions takes the existing function as an argument
#3. Outer function need to return inner function
```

```python
def Decorate_data(p): #p=Data
    def inner():
        p()
        print("hi")
    return inner

    
@ Decorate_data
def Data():
    print("Hello")

#Decorate the function data such that it displays Hello,hi
```

```python
Data()
```

```python
def Decorate_data1(p1):
    def inner1():
        p1()
        print("hii")
    return inner1


@ Decorate_data1
def Data1():
    print("Hola")
```

```python
def Decorate_n(q): #q=number()
    def inner2():
        p=q()
        return p**2
    return inner2

@ Decorate_n
def number():
    n = 10  # demo value (Kaggle-safe: was interactive int(input(...)))
    return n
#Decorate the function number such that it computes the square of a number
```

```python
number()
```

```python
def square(num):
    def inner():
        p=num()
        return p**2
    return inner


@square
def number():
    n = 10  # demo value (Kaggle-safe: was interactive int(input(...)))
    return n
```

```python
print("Square= ",number())
```

```python
def dec_squares(s):
    def sum_of_squares():
        n1,n2=s()
        return n1**2+n2**2
    return sum_of_squares
        

@ dec_squares
def data():
    n1 = 10  # demo value (Kaggle-safe: was interactive int(input(...)))
    n2 = 25  # demo value (Kaggle-safe: was interactive int(input(...)))
    return n1,n2
#Decorate function data to compute the sum of squares of two numbers read from user
```

```python
print("Sum of Squares= ",data())
```

```python
'''Given a list of Dictionaries where each dictionary contains sales data for a store.
Each dictionary includes the store name, the month and the sales figure for that month.
your task is to analyse these data to answer ther following questions
1. Calculate the total sale for each store across all months
2. Determine the month in which each store have the highest sales
3. Find the store with the highest total sales and the amount of those sales'''
```

```python
sales_data = [
    {"store": "Store A", "month": "January", "sales": 15000},
    {"store": "Store A", "month": "Febraury", "sales": 17000},
    {"store": "Store B", "month": "January", "sales": 12000},
    {"store": "Store B", "month": "Febraury", "sales": 13000},
    {"store": "Store C", "month": "January", "sales": 20000},
    {"store": "Store C", "month": "Febraury", "sales": 19000},
]
```

```python
def total_sales(sales_data):
    """Q1: total sales per store, across all months."""
    totals = {}
    for row in sales_data:
        store = row['store']
        totals[store] = totals.get(store, 0) + row['sales']
    return totals

def best_month_per_store(sales_data):
    """Q2: the month with the highest sales, for each store."""
    best = {}
    for row in sales_data:
        store, month, sales = row['store'], row['month'], row['sales']
        if store not in best or sales > best[store][1]:
            best[store] = (month, sales)
    return best

def top_store(totals):
    """Q3: the store with the highest total sales, and the amount."""
    return max(totals.items(), key=lambda kv: kv[1])

totals = total_sales(sales_data)
print("Total sales per store:", totals)

best = best_month_per_store(sales_data)
print("Best month per store:", best)

top_store_name, top_store_sales = top_store(totals)
print(f"Top store: {top_store_name} with total sales of {top_store_sales}")
```
