# Data Visualization: Histograms, Pie Charts & More

A tour of common matplotlib/seaborn chart types used in EDA.

**Covers:**
- Histograms for distribution
- Pie charts for proportions
- Box plots for outlier detection
- Distribution plots (`sns.histplot`)
- Correlation heatmaps

```python
#Histogram
#show distribution data
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
samples=np.random.randn(1000)
plt.hist(samples,bins=30,color='green')
plt.title('Histogram')
plt.xlabel('Value')
plt.ylabel('frequency')
plt.show()
```

```python
#pie chart
#Circular statistical graph which is divided into slices to show
#numerical proportions of data
sizes=[15,30,45,10]
labels=['A','b','C','D']
colors=['gold','green','yellow','red']
plt.pie(sizes,labels=labels,colors=colors,autopct='%1.1f%%',startangle=140)
plt.title('Pie chart')
plt.show()
```

```python
#Box plot
#detect outliers
#quartiles--- q1,q2,q3
data=[np.random.randn(100)]
```

```python
plt.boxplot(data,vert=True)
plt.title('Box Plot')  # fixed: title() needs a label argument
plt.show()
```

```python
data=pd.DataFrame({'category':['A','A','B','B','C','C','D','D'],
                  'values':[10,15,20,25,30,35,40,45]})
```

```python
#distpllot
#visualize univariate distribution
sns.displot(data['values'])
plt.title('Distribution plot')
plt.show()
```

```python
#heatmap
#used for representation correlation
#used to plot rectangular data as a color encoded matrix
correlation = data.corr(numeric_only=True)  # numeric_only=True: skip the non-numeric 'category' column (fixed for pandas 2.x+)
sns.heatmap(correlation,annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
