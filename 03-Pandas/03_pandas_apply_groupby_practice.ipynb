# Pandas: Apply & GroupBy Practice

Applied practice with custom functions, `.apply()`, and `groupby()` on structured data.

**Covers:**
- Sorting, grouping, and aggregating data
- Applying custom functions across columns

```python
import numpy as np
import pandas as pd
```

```python
d3=pd.DataFrame({'color':['red','blue','red','red','yellow'],
                'price':[100,200,100,300,400]}) 
```

```python
d3['color'].unique()
```

```python
d3['color'].value_counts()
```

```python
d3['price'].sum()
```

```python
d3['price'].mean()
```

```python
def sample(s):
    return s*0.5
```

```python
print(d3)
```

```python
#apply --- used to apply user defined function to dataframea
d3['price'].apply(sample)
```

```python
d3.sort_values(by='price',inplace=True)  # by default ascending order
```

```python
print(d3)
```

```python
d3.sort_values(by='price',inplace=True,ascending=False) #ascending=False gives descending order
```

```python
print(d3)
```

```python
data = pd.DataFrame({'customer':['c1','c1','c2','c2','c3','c3'],
                    'product1':[1,2,3,4,5,6],
                    'product2':[10,20,30,40,50,60]})
```

```python
print(data)
```

```python
#groupby
d=data.groupby('customer')
```

```python
print(data)
```

```python
d.mean()
```
