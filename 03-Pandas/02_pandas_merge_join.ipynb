# Pandas: Merge, Join & Column Operations

Combining DataFrames and common column-level operations.

**Covers:**
- `merge()` (inner/outer/left/right) vs. `join()` (index-based)
- Column methods: `.unique()`, `.value_counts()`, `.apply()`, `.sort_values()`
- `groupby()` for aggregation
- Reading/writing CSV and Excel files

```python
#Pandas
import pandas as pd
import numpy as np
```

```python
#combining data frames
#Merge and join operator
#Merge ---combines dataframe using a column's values.
#join --- combines dataframe using index. (using common row label(inner,outer,left,right join))
d1 = pd.DataFrame({'A': [1,2,3], 'B':[100,200,300], 'C': [1,90,10]}, index = ['r0','r1','r2'])
```

```python
d2 = pd.DataFrame({'A': [1,2,33], 'C':[10,6,9],'E': [1,2,4] }, index = ['r0','r1','r2'])
```

```python
print(d1)
```

```python
print(d2)
```

```python
#merge 
pd.merge(d1,d2, how = 'inner',on = 'A')
```

```python
pd.merge(d1,d2, how = 'outer',on = 'A')
```

```python
pd.merge(d1,d2, how = 'right',on = 'A')
```

```python
pd.merge(d1,d2, how = 'left',on = 'A')
```

```python
d1 = pd.DataFrame({'Q1' : [1,2,3],'Q2' : [1,3,5]},index = ['r1','r2','r3'])
```

```python
d2 = pd.DataFrame({'Q3' : [1,2,3], 'Q4' : [1,2,3]}, index = ['r1','r2','r3'])
```

```python
d1.join(d2, how = 'inner')
```

```python
d3 = pd.DataFrame({'color' : ['red','blue','red','red', 'yellow'], 'price' : [100,200,100,300,400]})
```

```python
print(d3)
```

```python
d3
```

```python
d3['color'].unique()
```

```python
d3['color'].value_counts
```

```python
d3['price'].sum()
```

```python
d3['price'].mean()
```

```python
def sample(s):
    return s*0.5
```

```python
#apply --- used to apply a user defined function to dataframe
d3['price'].apply(sample)
```

```python
d3.sort_values(by = 'price', inplace = True) # If we don't use inplace then new dataframe will be created.
#The default values of this attribute is False and it returns the copy of the object.
print(d3)
```

```python
d3.sort_values(by = 'price', inplace = True, ascending= 0) 
print(d3)
```

```python
#grouping
data = pd.DataFrame({'customer' : ['c1', 'c1', 'c2', 'c2', 'c3', 'c3'], 'product1' : [1,2,3,4,5,6], 'proudct2' : [10,20,30,40,50,60]})
```

```python
print(data)
```

```python
d = data.groupby('customer')
```

```python
print(d)
```

```python
d.mean()
```
