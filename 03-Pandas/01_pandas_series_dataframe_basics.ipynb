```python
import pandas as pd
import numpy as np
```

```python
#creating series
#from list
#from arrays
```

```python
#from list
list1 = [1,2,3,4]
array1 = np.array(list1)
```

```python
s1 = pd.Series(data = list1)
s2 = pd.Series(data = array1)
```

```python
labels = ['r0','r1','r3','r4']
lb = ['1','2','3','4']
s3 = pd.Series(data = list1,index = labels)
s4 = pd.Series(s2,lb)
#s3 = pd.Series(list1,labels)
```

```python
#Access data from series
#row labels
print(s3)
print(s4)
```

```python
print(s3['r3'])
```

```python
s3[2] = 100
print(s3)
```

```python
data1 = pd.Series(data=[1,2,3],index = ['a','b','c'])
data2 = pd.Series(data = [1,2,3,4], index = ['a','f','c','b'])
```

```python
print(data1)
```

```python
print(data2)
```

```python
#Linear algreba
#we can add but same label will do one time

print(data1 + data2)
```

```python
print(data1 - data2)
```

```python
print(data1 * data2)
print()
print(data1 / data2)
```

```python
#combining of series
#Row wise (axix = 0)
#column wise (axis = 1)
result1 = pd.concat([data1,data2] ,axis = 0)
print(result1)
```

```python
#column wise (axis = 1)
result2 = pd.concat([data1,data2] ,axis = 1)
print(result2)
```

```python
type(result2)
```

```python
#Creating a dataframe without adding series
d1 = pd.DataFrame([[1,2,3],[4,5,6]])
print(d1)
```

```python
#from
d2 = pd.DataFrame(np.random.randn(3,2))
print(d2)
```

```python
d3 = pd.DataFrame([[1,2,3],[4,5,6],[7,8,9]],index = ['r0','r1','r3'], columns = ['c0','c1','c2'])
```

```python
print(d3)
```

```python
#Access data
#Access individual series
print(d3['c2'])
```

```python
#Access multiple series --dataFrame_name[[]]
print(d3[['c2','c1']])
```

```python
#access multiple row from data frame ---dataFrame_name.loc[[]]
print(d3.loc[['r1','r0']])
```

```python
d3.shape
```

```python
d3["c3"] = [10,20,30]
```

```python
print(d3['c3'])
```

```python
#remove column c1 from d3
#drop - returns a new DataFrame by removinf row/col
#if we use the drop it will not gonna delete from the orginal dataframe
d3.drop('c2',axis = 1)
```

```python
#remove c1 from original DataFrame
d3.drop('c2',axis = 1,inplace = True)
```

```python
print(d3)
```

```python
d3.drop('r1',axis = 0,inplace = True)
```

```python
print(d3)
```
