# NumPy: Array Basics

Core NumPy array creation and manipulation.

**Covers:**
- Array creation and basic attributes
- Unique values
- Random sampling

```python
#Day3
import numpy as np
```

```python
v1 = np.array(['blue','red','green','black', 'red','yellow','pink','black'])
```

```python
#to identify the unique element
print(np.unique(v1))
```

```python
#generate sample
#rand  --> generates samples that follows uniform distribution. 
#randn -->generates samples that follows normal distribution.
m1 = np.random.rand(3,2)
m2 = np.random.randn(4,3)
```

```python
print(m1,"\n",m2)
```

```python
#mean

m= np.mean(m1)
#resrict the decimal 
print(f"Mean =  {m:.2f}".format(m))
print(np.mean(m2))
```

```python
#standard deviation 
print(np.std(m1))
```

```python
#max
#min
print(np.max(m1))
print(np.min(m2))
```

```python
#Access data
#indexing
#slicing
v2 = np.random.randn(19)
```

```python
print(v2)
```

```python
print(v2[0])
```

```python
print(v2[0:5:2])
```

```python
v2[0]= 234.618
#array is mutable
print(v2)
```

```python
#replacing single and spicified index value
v2[0:5:2] = [12.5,13.5,14.5]
print(v2)
```

```python
print(m1[0,1])
```

```python
print(m1[0:2,0:2])
```

```python
m1[0,0] = 3.45
print(m1)
```

```python
m3 = np.array([[100,200],[300,400]])
print(m3)
```

```python
#change the elements of first row ---- 1000 2000
sub=m3[0:1,0:2] = [1000,2000]
print(sub)
```

```python
#By default any changes done to sub matrix will refelect the orginal matrix as well
su = m3[0:1,0:2]
su[0] = 1000
print(su)
```

```python
m4 = np.array([[1,2],[3,4]])
print(m4)
```

```python
#IF we change the data it will not refelct in the orignal matrix ---.copy()
s1 = m4[1:2,0:2].copy()
print(s1)
```

```python
s1[:] = 100
print(s1)
print()
print(m4)
```

```python
#save single np array to a file ---.npy
#save mutiple np arrays to a file ---.npz
np.save('array1',m1) #save method is meant for saving single array to a .npy file
np.savez('array2',f1 = v1,f2 = m1, f3 =m2)
```

```python
#load ---- is used to retrieve contents from npy and npz files
data1 = np.load('array1.npy')
print(data1)
```

```python
data2 = np.load('array2.npz')
print(data2)
```

```python
print(data2['f1'])
print(data2['f2'])
print(data2['f3'])
```
